```python
import numpy as np
import pandas as pd
import sys

sys.path.append('../')  # allows notebook to import from src/

from src.neural_net import NeuralNetworkScratch

print("Import successful")
```

```python
df = pd.read_csv('data/final_prepared.csv')

print("Columns:", df.columns.tolist())
print("Shape:", df.shape)
print("First row preview:")
print(df.head(2))
```

```python
# Drop the target column to get features
X = df.drop('median_house_value', axis=1).values  # shape: (n_samples, n_features)
y = df['median_house_value'].values.reshape(-1, 1) # shape: (n_samples, 1)

print("X shape:", X.shape)   # e.g. (20640, 9)
print("y shape:", y.shape)   # e.g. (20640, 1)
print("Sample X row:", X[0])
print("Sample y value:", y[0])
```

```python
input_size = X.shape[1]  # auto-detects feature count from Week 1 data

nn = NeuralNetworkScratch(input_size=input_size)

# Run the forward pass
y_pred, cache = nn.forward(X)

print("Prediction shape:", y_pred.shape)        # must be (n_samples, 1)
print("Sample predictions:", y_pred[:5].flatten())
print("Actual values:     ", y[:5].flatten())
```

```python
# MSE before training — will be very large since weights are random
mse_untrained = np.mean((y_pred - y) ** 2)
print(f"MSE (untrained, expected to be large): {mse_untrained:.2f}")

# Critical checks — if these fail, something is wrong with the data or weights
print("Any NaN in predictions?", np.any(np.isnan(y_pred)))   # must be False
print("Any Inf in predictions?", np.any(np.isinf(y_pred)))   # must be False
```

```python
# Unpack cache — Member 2 will use all these values for backpropagation
X_in, Z1, A1, Z2, A2, Z3 = cache

print("Cache shapes — Member 2 will use these for backprop:")
print(f"  X  (input)         : {X_in.shape}")   # (n, input_size)
print(f"  Z1 (pre-relu L1)   : {Z1.shape}")     # (n, 64)
print(f"  A1 (post-relu L1)  : {A1.shape}")     # (n, 64)
print(f"  Z2 (pre-relu L2)   : {Z2.shape}")     # (n, 32)
print(f"  A2 (post-relu L2)  : {A2.shape}")     # (n, 32)
print(f"  Z3 (output)        : {Z3.shape}")     # (n, 1)

# Verify relu_derivative works (needed by Member 2)
test_Z = np.array([[-2.0, 0.0, 3.0]])
print("\nrelu_derivative test (expect [0, 0, 1]):", nn.relu_derivative(test_Z))
```
