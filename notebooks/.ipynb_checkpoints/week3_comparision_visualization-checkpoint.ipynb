```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
batch_loss = np.load("../results/batch_loss.npy")
sgd_loss = np.load("../results/sgd_loss.npy")
minibatch_loss = np.load("../results/minibatch_loss.npy")

print("Batch GD epochs:", len(batch_loss))
print("SGD epochs:", len(sgd_loss))
print("Mini-Batch GD epochs:", len(minibatch_loss))
```

```python
plt.figure(figsize=(10,6))

plt.plot(batch_loss, label="Batch GD", linewidth=2)

plt.plot(sgd_loss, label="SGD", linestyle="--")

plt.plot(minibatch_loss, label="Mini-Batch GD", linestyle="-.")

plt.xlabel("Epochs")

plt.ylabel("MSE")

plt.title("Comparision of Gradient Descent Variants")

plt.legend()

plt.grid(True)

plt.show()
```

```python
comparision = pd.DataFrame({
    "Variant": ["Batch GD", "SGD", "Mini-Batch GD"],

    "Final Loss": [
        batch_loss[-1],
        sgd_loss[-1],
        minibatch_loss[-1]
    ],

    "Speed": [
        "Slow (full dataset)",
        "Fast (per sample)",
        "Moderate (per batch)"
    ],

    "Stability": [
        "High",
        "Low (noisy)",
        "Medium"
    ],

    "Accuracy": [
        "Good baseline",
        "Fluctuates",
        "Balanced"
    ]
})

print(comparision)

comparision.to_csv("../results/week3_comparision.csv", index=False)
```
