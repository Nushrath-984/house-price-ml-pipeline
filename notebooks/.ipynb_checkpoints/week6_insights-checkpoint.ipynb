```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import Ridge

# Load training data
X_train = np.load('../results/X_train.npy')
y_train = np.load('../results/y_train.npy')

# Train Ridge model with best alpha
model_lr = Ridge(alpha=10)
model_lr.fit(X_train, y_train)

# Coefficients = feature importance
importance = model_lr.coef_.flatten()

plt.figure(figsize=(10,5))
plt.bar(range(len(importance)), importance)
plt.title('Feature Importance - Linear Regression')
plt.xlabel('Feature Index')
plt.ylabel('Coefficient Value')
plt.grid(True)
plt.savefig('../results/plots/feature_importance_lr.png')
plt.show()

print("✅ Feature importance plot saved!")
```

```python
import sys
sys.path.append('../src')

import pandas as pd
from sklearn.metrics import mean_squared_error
from neural_net import NeuralNetworkScratch

X_train = np.load('../results/X_train.npy')
y_train = np.load('../results/y_train.npy')

baseline_model = NeuralNetworkScratch(input_size=X_train.shape[1], learning_rate=0.0001)
baseline_model.fit(X_train, y_train, epochs=30)
baseline_mse = mean_squared_error(y_train, baseline_model.predict(X_train))

ablation_results = []
for i in range(X_train.shape[1]):
    X_ablate = np.delete(X_train, i, axis=1)
    model = NeuralNetworkScratch(input_size=X_ablate.shape[1], learning_rate=0.0001)
    model.fit(X_ablate, y_train, epochs=30)
    mse = mean_squared_error(y_train, model.predict(X_ablate))
    ablation_results.append((i, mse))

df_ablation = pd.DataFrame(ablation_results, columns=['FeatureIndex','MSE'])
df_ablation.to_csv('../results/week6_ablation_nn.csv', index=False)

print("✅ Ablation study completed and saved as week6_ablation_nn.csv")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score

X_tr, X_val, y_tr, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=42)

model = NeuralNetworkScratch(input_size=X_tr.shape[1], learning_rate=0.0001)
model.fit(X_tr, y_tr, epochs=30)

train_r2 = r2_score(y_tr, model.predict(X_tr))
val_r2 = r2_score(y_val, model.predict(X_val))

print("Train R²:", train_r2)
print("Validation R²:", val_r2)

if train_r2 - val_r2 > 0.1:
    print("⚠️ Model may be overfitting — consider adding L1/L2 regularization.")
else:
    print("✅ Model generalizes well.")
```
