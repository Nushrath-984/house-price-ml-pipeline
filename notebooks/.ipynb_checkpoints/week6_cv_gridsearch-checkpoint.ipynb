```python
import numpy as np, pandas as pd
from sklearn.model_selection import KFold
from sklearn.metrics import mean_squared_error, r2_score

# Load training data (from preprocessing outputs)
X = np.load('../results/X_train.npy')
y = np.load('../results/y_train.npy')

kf = KFold(n_splits=5, shuffle=True, random_state=42)

def evaluate_model(model_class, X, y, **kwargs):
    mse_scores, r2_scores = [], []
    for train_idx, val_idx in kf.split(X):
        X_train, X_val = X[train_idx], X[val_idx]
        y_train, y_val = y[train_idx], y[val_idx]
        
        model = model_class(**kwargs)
        model.fit(X_train, y_train)
        y_pred = model.predict(X_val)
        
        mse_scores.append(mean_squared_error(y_val, y_pred))
        r2_scores.append(r2_score(y_val, y_pred))
    return np.mean(mse_scores), np.mean(r2_scores)
```

```python
from sklearn.linear_model import Ridge

alphas = [0.01, 0.1, 1, 10, 100]
results_lr = []
for a in alphas:
    mse, r2 = evaluate_model(Ridge, X, y, alpha=a)
    results_lr.append((a, mse, r2))

df_lr = pd.DataFrame(results_lr, columns=['Alpha','MSE','R2'])
df_lr.to_csv('../results/week6_lr_gridsearch.csv', index=False)
print(df_lr)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X = scaler.fit_transform(X)
```

```python
import sys
sys.path.append('../src')
from neural_net import NeuralNetworkScratch
model = NeuralNetworkScratch(input_size=X.shape[1], learning_rate=0.0001)
model.fit(X, y, epochs=30)
y_pred = model.predict(X)
print(np.mean((y - y_pred.flatten())**2))
```

```python
import sys
sys.path.append('../src')
from neural_net import NeuralNetworkScratch

batch_sizes = [16, 32, 64]
layers = [1, 2, 3]
lrs = [0.0001, 0.001]

results_nn = []
for bs in batch_sizes:
    for l in layers:
        for lr in lrs:
            mse, r2 = evaluate_model(NeuralNetworkScratch, X, y,
                                     input_size=X.shape[1],
                                     batch_size=bs, num_layers=l, learning_rate=lr)
            results_nn.append((bs, l, lr, mse, r2))

df_nn = pd.DataFrame(results_nn, columns=['BatchSize','Layers','LR','MSE','R2'])
df_nn.to_csv('../results/week6_nn_gridsearch.csv', index=False)
print(df_nn.head())
```

```python
best_lr = df_lr.loc[df_lr['R2'].idxmax()]
best_nn = df_nn.loc[df_nn['R2'].idxmax()]

best_lr.to_csv('../results/week6_best_lr.csv', index=False)
best_nn.to_csv('../results/week6_best_nn.csv', index=False)

print("Best Linear Regression:", best_lr.to_dict())
print("Best Neural Net:", best_nn.to_dict())
```
